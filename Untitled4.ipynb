### 5. Identifying Features (Independent Variables) and Target (Dependent Variable)

Based on our exploration of the `Boutique_Store_300_Data_cleaned.csv` dataset, here is how the variables are classified for predictive modeling or regression analysis:

#### **Dependent Variable (Target)**
* **`Total Sale (₱)`**: This is typically the target variable we want to predict or analyze. It represents the overall revenue generated from a transaction.

#### **Independent Variables (Features)**
These are the attributes used to predict, explain, or model the target variable:
* **Direct Numerical Predictors**:
  * `Unit Price (₱)` (Base cost of a single unit of the product)
  * `Quantity` (Number of units purchased)
  * `Discount (%)` (Discount rate applied to the sale)
* **Categorical Features**:
  * `Category` (e.g., Top, Pants, Footwear)
  * `Product Name` (Specific product identity)
  * `Size` (e.g., S, M, L, One Size)
  * `Color` (e.g., White, Black, Beige)
  * `Payment Method` (e.g., Cash, GCash, Maya, Debit Card)
* **Temporal Features**:
  * `Date` (Can be decomposed into Month, Day of the Week, or Seasonality)
* **Identifiers (Usually Excluded from Modeling)**:
  * `Sale ID` (Unique transactional identifier)
  * `Customer Name` (Customer identifier)

### 6. Splitting the Dataset into Training and Testing Sets
We will now prepare our features ($X$) and target ($y$) for modeling. Since our dataset contains categorical features (like `Category`, `Size`, and `Payment Method`), we will convert them into numerical form using one-hot encoding. Then, we will split the dataset into 80% training and 20% testing sets.

```python
from sklearn.model_selection import train_test_split

# Define our initial feature set and target variable
# We exclude unique identifiers ('Sale ID', 'Customer Name', 'Date') from the predictive features
feature_cols = ['Product Name', 'Category', 'Size', 'Color', 'Unit Price (₱)', 'Quantity', 'Discount (%)', 'Payment Method']
X = df[feature_cols]
y = df['Total Sale (₱)']

# One-hot encode categorical variables
X_encoded = pd.get_dummies(X, columns=['Product Name', 'Category', 'Size', 'Color', 'Payment Method'], drop_first=True)

# Split into training and testing sets (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# Display the shapes of the resulting splits
print(f"Training set features shape: {X_train.shape}")
print(f"Testing set features shape: {X_test.shape}")
print(f"Training set target shape: {y_train.shape}")
print(f"Testing set target shape: {y_test.shape}")
```

### 7. Training a Decision Tree Classifier
To build a **Decision Tree Classifier**, we need a categorical target. We will define a new binary target variable:
* **`1` (High Value Sale)**: If `Total Sale (₱)` is strictly above the median value.
* **`0` (Standard Value Sale)**: If `Total Sale (₱)` is at or below the median value.

Then, we will split this updated classification dataset and train our classifier.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Define binary target based on median sales threshold
median_sale = y.median()
y_class = (y > median_sale).astype(int)

# Split the dataset again using the new classification target
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_encoded, y_class, test_size=0.2, random_state=42
)

# Initialize and train the Decision Tree Classifier
clf = DecisionTreeClassifier(max_depth=4, random_state=42)
clf.fit(X_train_c, y_train_c)

# Predict on the test set
y_pred = clf.predict(X_test_c)

# Evaluate the model
accuracy = accuracy_score(y_test_c, y_pred)
print(f"Median sales threshold: ₱{median_sale:.2f}\n")
print(f"Model Accuracy: {accuracy * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test_c, y_pred, target_names=['Standard Value', 'High Value']))
```

### 8. Train the Model Using the Training Dataset
We will now explicitly fit our `DecisionTreeClassifier` model to the training subset (`X_train_c` and `y_train_c`) to learn the decision rules for classifying sales.

```python
# Train the Decision Tree Classifier model
clf.fit(X_train_c, y_train_c)

print("Model training completed successfully!")
print(f"Number of training samples used: {len(X_train_c)}")
print(f"Number of features used for training: {X_train_c.shape[1]}")
```

### 9. Visualizing the Decision Tree Structure
We will use `sklearn.tree.plot_tree` to visualize our trained decision tree classifier. This helps us interpret the hierarchical rules the model uses to categorize transactions.

```python
from sklearn.tree import plot_tree

# Set a clean figure size
plt.figure(figsize=(24, 12), dpi=150)

# Plot the decision tree
plot_tree(
    clf,
    feature_names=list(X_train_c.columns),
    class_names=['Standard Value', 'High Value'],
    filled=True,
    rounded=True,
    fontsize=10
)

plt.title("Decision Tree Classifier Structure (Predicting High Value Sales)", fontsize=16, fontweight='bold')
plt.show()
```

### 10. Generate Predictions on the Testing Dataset
We will now use our trained model to make predictions on the unseen testing dataset (`X_test_c`) and compare the predicted outcomes with the actual ground truth targets.

```python
# Generate predictions using the testing dataset
y_test_pred = clf.predict(X_test_c)

# Create a comparison DataFrame for display
comparison_df = pd.DataFrame({
    'Actual Value': y_test_c.values,
    'Predicted Value': y_test_pred
})

# Map the 0 and 1 categories back to readable labels
label_map = {0: 'Standard Value', 1: 'High Value'}
comparison_df['Actual Label'] = comparison_df['Actual Value'].map(label_map)
comparison_df['Predicted Label'] = comparison_df['Predicted Value'].map(label_map)

# Display the first 10 predictions vs actuals
print("Sample of Test Set Predictions vs Actual Targets:")
display(comparison_df.head(10))
```

### 11. Calculate and Display Model Accuracy
We will now explicitly calculate and display the accuracy score of our Decision Tree Classifier on the testing dataset.

```python
from sklearn.metrics import accuracy_score

# Calculate accuracy
test_accuracy = accuracy_score(y_test_c, y_test_pred)

print("--- Model Accuracy Report ---")
print(f"Accuracy Score: {test_accuracy:.4f}")
print(f"Accuracy (Percentage): {test_accuracy * 100:.2f}%")
```

### 12. Generate a Confusion Matrix
We will now generate and plot a confusion matrix using `scikit-learn` to see how well our classifier's predictions (Standard vs. High Value) align with the actual values.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Compute the confusion matrix
cm = confusion_matrix(y_test_c, y_test_pred)

# Setup the display
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Standard Value', 'High Value'])

# Plot the confusion matrix
fig, ax = plt.subplots(figsize=(8, 6))
disp.plot(cmap='Blues', ax=ax, values_format='d')
plt.title('Confusion Matrix: Decision Tree Classifier', fontsize=14, fontweight='bold')
plt.grid(False) # Turn off grid lines for clean plotting
plt.show()
```

### 13. Display the Classification Report
We will now generate and display the metrics (Precision, Recall, F1-Score, and Support) to evaluate the classifier's performance on each class.

```python
from sklearn.metrics import classification_report

# Generate classification report
report = classification_report(y_test_c, y_test_pred, target_names=['Standard Value', 'High Value'])

print("--- Classification Report ---")
print(report)
```

### 14. Feature Importances (Variable Influence)
We will extract and visualize the feature importances from our trained Decision Tree Classifier to identify which variables have the greatest influence on predicting sales value classifications.

```python
import numpy as np

# Extract feature importances
importances = clf.feature_importances_
feature_names = X_train_c.columns

# Sort them in descending order
indices = np.argsort(importances)[::-1]

# Create a DataFrame for visualization
importance_df = pd.DataFrame({
    'Feature': feature_names[indices],
    'Importance': importances[indices]
}).head(10) # Look at top 10 most influential features

print("--- Top Feature Importances ---")
display(importance_df)

# Plot top feature importances
plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', data=importance_df, palette='viridis', hue='Feature', legend=False)
plt.title('Top 10 Most Influential Features (Decision Tree)', fontsize=14, fontweight='bold')
plt.xlabel('Feature Importance Score')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```
